```python
import cv2
import numpy as np
import tensorflow as tf
from ultralytics import YOLO
from ultralytics import YOLO
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
face_model=YOLO(r"D:\projects\facec detection\Face_detection\Face detection_enhanced\train\weights\best.pt")
from tensorflow.keras.models import load_model
emotion_model=load_model("best_emotion_model_scratch_enhanced.keras",custom_objects={"preprocess_input":preprocess_input})     
# ==============================
# Load Models
# ==============================

# أسماء المشاعر (غيرهم حسب ترتيب تدريبك)
emotions = ["angry", "disgust", "fear", "happy", "neutral","sad","surprise"]

# ==============================
# Start Camera
# ==============================

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # ==============================
    # Face Detection
    # ==============================

    results = face_model(frame)

    for box in results[0].boxes.xyxy:
        x1, y1, x2, y2 = map(int, box)

        # Crop face
        face = frame[y1:y2, x1:x2]

        if face.size == 0:
            continue

        # ==============================
        # Preprocess for Emotion Model
        # ==============================

        face = cv2.resize(face, (48, 48))  # غير المقاس لو موديلك مختلف
        #face = cv2.cvtColor(face, cv2.COLOR_BGR2GRAY)
        face = face / 255.0
        face = np.reshape(face, (1, 48, 48, 3))

        # ==============================
        # Emotion Prediction
        # ==============================

        prediction = emotion_model.predict(face, verbose=0)
        emotion_label = np.argmax(prediction)
        confidence = np.max(prediction)

        emotion_text = f"{emotions[emotion_label]} ({confidence:.2f})"

        # ==============================
        # Draw Results
        # ==============================

        y_text = y1 - 10 if y1 - 10 > 10 else y1 + 20

        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

        (text_width, text_height), _ = cv2.getTextSize(
            emotion_text, cv2.FONT_HERSHEY_SIMPLEX, 0.7, 2
        )

        cv2.rectangle(
            frame,
            (x1, y_text - text_height - 5),
            (x1 + text_width, y_text + 5),
            (0, 255, 0),
            -1,
        )

        cv2.putText(
            frame,
            emotion_text,
            (x1, y_text),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 0, 0),
            2,
        )

    # ==============================
    # Show Frame
    # ==============================

    cv2.imshow("Live Face & Emotion Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
