```python
import os
import cv2
from matplotlib import pyplot as plt
import glob

# ===== مسار فولدر نتائج YOLO =====
results_dir = r"D:\projects\facec detection\Face_detection\Face detection_enhanced\train"

# ===== جلب كل الصور في الفولدر recursively =====
all_images = glob.glob(os.path.join(results_dir, "**/*.jpg"), recursive=True)

for img_path in all_images:
    img = cv2.imread(img_path)
    if img is None:
        continue
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    plt.imshow(img)
    plt.title(os.path.basename(img_path))
    plt.axis("off")
    plt.show()
```

```python
for img_path in all_images:
    img = cv2.imread(img_path)
    if img is None:
        continue
    cv2.imshow("Images", img)
    
    if cv2.waitKey(10000) & 0xFF == ord('q'):  # كل صورة تظهر نص ثانية
        break

```
