```python
import cv2
from ultralytics import YOLO

# 1️⃣ تحميل الموديل مع weights اللي انت حافظه
model = YOLO("Face detection_enhanced/train/weights/best.pt")  

# 2️⃣ تشغيل الكاميرا
cap = cv2.VideoCapture(0)  # 0 = الكاميرا الافتراضية

if not cap.isOpened():
    print("Error: Camera not accessible")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # 3️⃣ YOLOv8 inference على كل فريم
    results = model.predict(frame, conf=0.2, imgsz=640)  # confidence بين 0 و1

    # 4️⃣ رسم البوكسات يدويًا على الفريم
    for r in results:
        boxes = r.boxes.xyxy  # إحداثيات البوكس [x1, y1, x2, y2]
        scores = r.boxes.conf  # confidence
        class_ids = r.boxes.cls  # class index
        names = r.names  # dictionary للكلاسات {0: 'Rock', 1: 'Paper', ...}

        for i, box in enumerate(boxes):
            x1, y1, x2, y2 = map(int, box)
            conf = float(scores[i])
            class_id = int(class_ids[i])
            label = f"{names[class_id]} {conf:.2f}"

            # رسم rectangle
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            # رسم label
            cv2.putText(frame, label, (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # 5️⃣ عرض الفريم
    cv2.imshow("YOLOv8 Webcam", frame)

    # الخروج بالضغط على 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 6️⃣ تنظيف
cap.release()
cv2.destroyAllWindows()
```
